```python
#We create a reusable cleaning function
# \s ---> white spaces  \S ----> matches everything except the space

def clean_text(text):
    text = text.lower()  # lowercase
    text = re.sub(r'\S+@\S+', '', text)  # remove emails  #\b means start of the word and \b means end of the word
    text = re.sub(r'http\S+', '', text)  # remove URLs
    text = re.sub(r'[^a-zA-Z\s]', '', text)  # remove special chars
    text = re.sub(r'\s+', ' ', text).strip()  # remove extra spaces
    return text
```

```python
import nltk
import spacy
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
#downloading NLTK Data
#using this for tokenization,stopword removal,resume cleaning

nltk.download('punkt')
nltk.download('stopwords')
```

```python
#loading spacy model

nlp = spacy.load("en_core_web_sm")
```

```python
#define stopwords

stop_words = set(stopwords.words('english'))
```

```python
#advanced cleaning function

def advanced_cleaning(text):
    words = word_tokenize(text)  #tokenize means splitting text into individual words
    words = [word for word in words if word not in stop_words and len(word)>2]  # going through each word and removes common useless words and using 
    #len(word)>2 removes small words like a,an,it etcc....
    return " ".join(words)  #comibe back list to string 

#doing this to reduce noise ,keep meaningful words, impore ML model accuracy for our resume screening
    
```

```python
#making a new column and applying the def function

df["Processed_text"] = df["cleaned_text"].apply(advanced_cleaning)
```

```python
#lemmatization is done to convert a word to its base/root form (meaningful one)
#example : running ----> run, better ----> good, studies ---->study
# we are using lemmatization instead of stemming because its accuracy is high and used in NLP/ML models and also helps to reduce duplicate meaning
# and keeps meaningful words

def lemmatize_text(text):
    doc = nlp(text)  #storing NLP rich structured data in doc 
    return " ".join([token.lemma_ for token in doc])  #converts list to string
    
#token.lemma_
#Returns base form of word
```

```python
#spaCy:

#Tokenizes text
#Understands grammar
#Assigns lemma to each word

#we can see gramatically error but ML cares about meaning not grammar

#Avoid Lemmatization when:

#Grammar matters (like chat output)
# we are using this becuase developing,developer,developed etc.. in resume ------->develop using lemma
```

```python
df["Lemmatized_text"] = df["Processed_text"].apply(lemmatize_text)
```

```python
def extract_keywords(text):
    doc = nlp(text)  #process texts using spacy
    keywords = [token.text for token in doc if token.pos_ in ['NOUN', 'PROPN']]   #token.text --->original word, token.pos---->parts of speech
    return keywords


#Keep only:

#NOUN → general important terms
#PROPN → specific names (skills, tools, companies)

#returns Output = list of important words
```

```python
#What is the goal of this function?

#👉 To keep only important words like:

#skills (Python, SQL)
#roles (Analyst, Engineer)
#tools (Tableau, Power BI)
#names (companies, technologies)

#👉 And remove less useful words like:

#is, the, was, very, etc.


#example
#text = "I am a Data Analyst skilled in Python and SQL at Google"

#output
#['Data', 'Analyst', 'Python', 'SQL', 'Google']



#helps is resume screening,feature engineering,job matching
```

```python
df["Keywords"] = df["Lemmatized_text"].apply(extract_keywords)
```

```python
#role based skill dictionary 

role_skills = {
    "Data Analyst": {"python", "sql", "excel", "data", "analysis"},
    "Data Scientist": {"python", "machine", "learning", "model", "statistics"},
    "HR Analyst": {"hr", "recruitment", "employee", "analytics"},
}
```

```python
#weighted skill matching score 

def skill_match_score(jd_keywords, resume_keywords, role):
    jd_set = set(jd_keywords)
    resume_set = set(resume_keywords)
    
    role_specific = role_skills.get(role, set())
    
    score = 0
    
    for word in jd_set:
        if word in resume_set:
            if word in role_specific:
                score += 3   # high importance
            else:
                score += 1
    
    return score / (3 * len(jd_set))
```

```python
#experience score function

def experience_score(text):
    word_count = len(text.split())
    
    if word_count < 100:
        return 0.3
    elif word_count < 300:
        return 0.6
    else:
        return 1.0
```

```python
#final score function

def final_score(similarity, skill_score, exp_score):
    return (
        0.4 * similarity +
        0.5 * skill_score +
        0.1 * exp_score
    )  
```

```python
#master function

def rank_candidates(jd_index, top_n=5):
    results = []
    
    role = jd_df.iloc[jd_index]['role']  # get role once
    
    for i in range(len(df)):
        
        # --- STEP 1: HARD FILTER (VERY IMPORTANT) ---
        if role == "Data Analyst":
            resume_keywords = df.iloc[i]['Keywords']

            # Require at least 2 strong skills
            required_skills = ["python", "sql", "analysis", "data"]
            
            match_count = sum(1 for skill in required_skills if skill in resume_keywords)
            
            if match_count < 2:
                continue

        
        if role == "HR Analyst":
            resume_keywords = df.iloc[i]['Keywords']

            required_skills = ["hr", "recruitment", "employee", "management"]

            match_count = sum(1 for skill in required_skills if skill in resume_keywords)

            if match_count < 2:
                continue
        
        # --- STEP 2: GET SCORES ---
        sim = all_similarity_scores[jd_index][i]
        
        skill = skill_match_score(
            jd_df.iloc[jd_index]['jd_keywords'],
            df.iloc[i]['Keywords'],
            role
        )
        
        # Optional filter (keep it)
        if skill < 0.2:
            continue
        
        exp = experience_score(df.iloc[i]['Lemmatized_text'])
        
        score = final_score(sim, skill, exp)
        
        results.append((i, score))
    
    # --- STEP 3: SORT ---
    results = sorted(results, key=lambda x: x[1], reverse=True)
    
    # --- STEP 4: PRINT ---
    print(f"\nTop {top_n} candidates for role: {role}\n")
    
    for idx, score in results[:top_n]:
        print("Final Score:", round(score, 3))
        print("Category:", df.iloc[idx]['category'])
        print("Preview:", df.iloc[idx]['resume_text'][:150])
        print("-" * 60)
    
    return results
```

```python
#clean context
def clean_output(text):
    # Remove everything before actual explanation
    lines = text.split("\n")
    
    for line in lines:
        if len(line.strip()) > 20 and "candidate" in line.lower():
            return line.strip()
    
    # fallback
    return lines[-1].strip()
```

```python
#strong prompt

def build_prompt(context):
    return f"""
Explain in one sentence why this candidate is suitable for a Data Analyst role.

{context}
"""
```

```python
#llm function
def ask_llm(context):
    prompt = build_prompt(context)
    
    response = generator(
        prompt,
        max_new_tokens=30,   # 🔥 reduce further
        do_sample=False
    )
    
    return response[0]['generated_text'].strip()
```

```python
def generate_explanation(idx, role):
    skills = df.iloc[idx]['Keywords']
    
    # Remove duplicates while preserving order
    unique_skills = list(dict.fromkeys(skills))
    
    if role == "Data Analyst":
        required = ["python", "sql", "excel", "analysis", "data"]
        matched = [s for s in unique_skills if s in required]
        
        # Take only unique meaningful skills
        matched = list(dict.fromkeys(matched))[:3]
        
        return f"Candidate has strong skills in {', '.join(matched)} which are essential for Data Analyst role."
    
    elif role == "HR Analyst":
        required = ["hr", "recruitment", "employee", "management"]
        matched = [s for s in unique_skills if s in required]
        matched = list(dict.fromkeys(matched))[:3]
        
        return f"Candidate has relevant HR experience and skills in {', '.join(matched)}."
    
    else:
        return "Candidate has relevant experience and skills matching the job requirements."
```
